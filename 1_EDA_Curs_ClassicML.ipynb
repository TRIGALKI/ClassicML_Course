```python
# Расчет корреляции Спирмена для целевых переменных
correlations = df.corr(method='spearman')[targets].drop(targets)

# Топ-10 положительных и отрицательных корреляций для IC50
top_ic50 = correlations['IC50, mM'].sort_values(ascending=False)

plt.figure(figsize=(10, 8))
sns.barplot(x=pd.concat([top_ic50.head(10), top_ic50.tail(10)]),
            y=pd.concat([top_ic50.head(10), top_ic50.tail(10)]).index,
            palette='coolwarm')
plt.title('Топ-20 дескрипторов по корреляции Спирмена с IC50, mM')
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
import numpy as np

# Подготовка данных для поиска выбросов (используем только числовые признаки)
X = df.select_dtypes(include=[np.number])

# Инициализация и обучение Isolation Forest
iso = IsolationForest(contamination=0.05, random_state=42)
outliers = iso.fit_predict(X)

# Добавляем метку в датафрейм: -1 это выброс, 1 это норма
df['is_outlier'] = outliers

print(f"Обнаружено выбросов: {(outliers == -1).sum()}")

# Визуализация выбросов в пространстве двух первых компонент (для наглядности)
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(10, 6))
plt.scatter(X_pca[outliers == 1, 0], X_pca[outliers == 1, 1], c='blue', label='Normal', alpha=0.5)
plt.scatter(X_pca[outliers == -1, 0], X_pca[outliers == -1, 1], c='red', label='Outlier', alpha=0.8)
plt.title('Визуализация выбросов через PCA (Isolation Forest)')
plt.legend()
plt.show()
```

### (Isolation Forest)

Алгоритм **Isolation Forest** основан на необычном подходе: вместо того чтобы моделировать «нормальные» данные, он пытается **изолировать** аномалии.

#### Основная идея:
1. **Случайное разбиение**: Алгоритм выбирает случайный признак и случайное значение для разделения данных (строит решающее дерево).
2. **Изоляция**: Выбросы — это точки, которые сильно отличаются от большинства. Чтобы «отрезать» такую точку от остального набора данных, обычно требуется очень мало случайных разделений.
3. **Длина пути**: Чем меньше глубина дерева, на которой точка оказалась одна в листе, тем выше вероятность, что это выброс. Нормальные точки «кучкуются» вместе, и чтобы изолировать их по отдельности, нужно построить очень длинную цепочку разделений.

#### Почему мы выбрали именно его:
* **Многомерность**: В нашем датасете более 200 химических дескрипторов. Обычные методы (вроде правила 3-х сигм или Z-оценки) работают только для одного признака и не видят сложные аномалии в комбинации признаков.
* **Нелинейность**: Он не предполагает, что данные распределены нормально.

#### Интерпретация PCA-графика:
Так как мы не можем увидеть 200-мерное пространство, мы использовали **PCA (Метод главных компонент)**, чтобы «сплющить» данные до 2D. На графике синие точки — это те, что требуют длинных путей для изоляции (норма), а красные — те, что изолируются быстро (выбросы).
